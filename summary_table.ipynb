# Baselines

```python
import pandas as pd
import numpy as np

imputation = {'ARIMA':'ARIMA','INTERPOLATION':'Interpolation','FF':'Forward fill','NO_IMP':'Model-based'}
loss_type = {'bce':'BCE','mse':'MSE(logit)'}
```

```python
base_results = pd.read_csv("test_outputs/baseline_output.txt",sep=" ", header=None)

datas = {}
for _,row in base_results.iterrows():
    model = row[0]
    mse = row[1]
    mae = row[2]
    mape = row[3]
    mape2 = row[4]
    r2 = row[5]
    imp_type = imputation[row[6]]
    loss = loss_type[row[8]]

    key = (model,imp_type,loss) 
    if key not in datas:
        datas[key] = [[],[],[],[],[]]
    datas[key][0].append(mse)
    datas[key][1].append(mae)
    datas[key][2].append(mape)
    datas[key][3].append(mape2)
    datas[key][4].append(r2)

pd_data = []
for model,imp_type,loss in datas:
    key = (model,imp_type,loss)
    pd_data.append([model,imp_type,loss, "%.2f" % (np.asarray(datas[key][0]).mean() / 1e17)+ "\u00B1" + "%.2f" % (np.asarray(datas[key][0]).std() / 1e17),"%.2f" % (np.asarray(datas[key][1]).mean() / 1e8) + "\u00B1" + "%.2f" % (np.asarray(datas[key][1]).std() / 1e8),"%.2f" % np.asarray(datas[key][2]).mean()+ "\u00B1"+"%.2f" % np.asarray(datas[key][2]).std(),"%.2f" % np.asarray(datas[key][3]).mean()+ "\u00B1"+ "%.2f" % np.asarray(datas[key][3]).std(),"%.2f" % np.asarray(datas[key][4]).mean()+ "\u00B1"+ "%.2f" % np.asarray(datas[key][4]).std()])

```

```python
base_df = pd.DataFrame(pd_data,columns=['Model',"Imputation","Loss","MSE, x10^17","MAE, x10^8","MAPE_all","MAPE_large","R^2"])
```

# Transformers

```python
tr_results = pd.read_csv("test_outputs/log.txt",sep=" ", header=None)
```

```python
tr_results
```

```python

datas = {}
for _,row in tr_results.iterrows():
    model = row[0]+row[1]
    mse = row[3]
    mae = row[4]
    mape = row[5]
    mape2 = row[6]
    r2 = row[7]
    imp_type = imputation[row[11]]
    loss = loss_type[row[13]]
    L = row[8]
    heads = row[9]
    if row[2]:
        fft = "+"
    else:
        fft = "-"

    if row[10]:
        recurrent = "+"
    else:
        recurrent = "-"
    
    key = (model,imp_type,loss,L,heads,fft,recurrent) 
    if key not in datas:
        datas[key] = [[],[],[],[],[]]
    datas[key][0].append(mse)
    datas[key][1].append(mae)
    datas[key][2].append(mape)
    datas[key][3].append(mape2)
    datas[key][4].append(r2)

pd_data = []
for model,imp_type,loss,L,heads,fft,recurrent in datas:
    key = (model,imp_type,loss,L,heads,fft,recurrent)
    pd_data.append([model,imp_type,loss,L,heads,fft,recurrent, "%.2f" % (np.asarray(datas[key][0]).mean() / 1e17)+ "\u00B1" + "%.2f" % (np.asarray(datas[key][0]).std() / 1e17),"%.2f" % (np.asarray(datas[key][1]).mean() / 1e8) + "\u00B1" + "%.2f" % (np.asarray(datas[key][1]).std() / 1e8),"%.2f" % np.asarray(datas[key][2]).mean()+ "\u00B1"+"%.2f" % np.asarray(datas[key][2]).std(),"%.2f" % np.asarray(datas[key][3]).mean()+ "\u00B1"+ "%.2f" % np.asarray(datas[key][3]).std(),"%.2f" % np.asarray(datas[key][4]).mean()+ "\u00B1"+ "%.2f" % np.asarray(datas[key][4]).std()])

```

```python
enc_res = pd.DataFrame(pd_data,columns=['Model',"Imputation","Loss","Layers","Attn. heads","FFT","Recurrent embeddings","MSE, x10^17","MAE, x10^8","MAPE_all","MAPE_large","R^2"])
```
